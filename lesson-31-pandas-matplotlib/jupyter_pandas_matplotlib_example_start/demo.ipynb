```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("data/Property_Assessment_Data_(Current_Calendar_Year)_20251030.csv")

data.head()
```

```python
print(data.info())
print(f"(Rows, Columns) {data.shape}")

data.describe()
```

```python
data["Tax Class"].value_counts()

mask = (data["Tax Class"] == "Residential")# & (data["Assessed Value"] > 300000)

residential_data = data[mask]

residential_data.head()
```

```python
top_10_most_expensive = residential_data.sort_values(by="Assessed Value", ascending=False)[:10]

top_10_most_expensive
```

```python
idx = residential_data["Assessed Value"].idxmax()
most_expensive_house = residential_data.loc[idx]
most_expensive_house
```

```python
top_6_expensive = top_10_most_expensive[:6]

top_6_expensive["Property Address"] = top_6_expensive["House Number"].astype(int).astype(str) + ' ' + top_6_expensive["Street Name"]

property_names = top_6_expensive["Property Address"]
assessed_values = top_6_expensive["Assessed Value"]

plt.figure(figsize=(10,6))
plt.bar(property_names, assessed_values)
plt.title("Top 6 Most Expensive Homes")
plt.xlabel("Property Names")
plt.ylabel("Assessed Value")
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(property_names, assessed_values, marker='o', linestyle='-', color='orange')
plt.xlabel('Property Address')
plt.ylabel('Assessed Value')
plt.title('Top 6 Most Expensive Residential Properties - Line Chart')
plt.xticks(rotation=45, ha='right')  # Rotate x labels for better readability
plt.grid(True)
plt.show()
```

```python
downtown = data[data['Neighbourhood'] == 'DOWNTOWN']

plt.figure(figsize=(10, 6))
plt.scatter(downtown['Latitude'], downtown['Assessed Value'], alpha=0.5, color='purple')
plt.xlabel('Latitude')
plt.ylabel('Assessed Value')
plt.title('Downtown Properties: Assessed Value vs Latitude')
plt.show()
```
